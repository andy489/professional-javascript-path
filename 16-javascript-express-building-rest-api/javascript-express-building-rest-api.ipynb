### Initialize a Node.js Project & Install Express

```bash
# 1. Create project directory
mkdir myapi && cd myapi

# 2. Initialize Node.js project (creates package.json)
npm init -y

# 3. Install Express
npm install --save express

# 4. Install middleware dependencies
npm install --save body-parser   # parses incoming request bodies
npm install --save cors          # enables cross-origin resource sharing
npm install --save nodemon       # auto-restarts server on file changes

# 5. Create the entry point
touch index.js
```

**Configure `package.json`** to use nodemon for development:

```json
{
  "main": "index.js",
  "scripts": {
    "dev": "nodemon src/index.js"
  }
}
```

**Start the development server:**

```bash
npm run dev
```

### Constructing a Web App with Express & JavaScript

A minimal Express app structure in `index.js`:

```javascript
const express = require('express');
const bodyParser = require('body-parser');
const cors = require('cors');

const app = express();
const port = 3000;

// Configure middleware
app.use(bodyParser.json());
app.use(cors());

// Root endpoint — smoke test
app.get('/', (req, res) => {
  res.send('REST API is running');
});

// Start the server
app.listen(port, () => console.log(`Hello world app listening on port ${port}!`));
```

Run directly with Node (no auto-restart):

```bash
node index.js
```

Then visit `http://localhost:3000/` to confirm the server responds.

## Module 02: Setting Up the REST API Project <a id="module-02"></a>

This module covers what REST APIs are, their design principles, how to plan and structure a REST API project, and the options for storing data.

### Overview of the REST API

**REST** (REpresentational State Transfer) is an architectural style and approach for communication often used in web development and for systems to communicate.

A **REST API** is an interface used by two or more computer systems to exchange data securely via HTTP requests (`GET`, `PUT`, `POST`, `DELETE`) over the internet.

**Why REST APIs?**

- **Limited bandwidth** — HTTP calls are fast and lightweight, making them efficient when bandwidth is constrained
- **Ease of coding** — implementing a REST API service is simpler than SOAP-based services
- **Easy of integration** — any client can use REST API apps, even if not specifically designed for it
- **Cache** — REST APIs are easier to cache since the server is stateless and requests are processed individually

### REST API Concepts

| Concept | Definition |
|---|---|
| **Client** | Software on a user's computer or smartphone that initiates communication |
| **Server** | Offers a REST API as a means of access to the app's data or features |
| **Stateless** | The client sends all info needed per request; the server holds no session state |
| **Header** | Stores info relevant to both client & server (API key, IP address, response format) |
| **Body** | Conveys additional info to the server (e.g. data to add or replace) |
| **Endpoint** | A URI indicating where and how to find the resource (e.g. a URL) |
| **Cacheable** | Response data must be labeled as cacheable or non-cacheable |
| **Uniform Interface** | A consistent interface that acts as a contract between client and service |
| **Layered System** | Each component communicates only with its immediate neighbour |
| **Code On Demand** | Client functionality can be extended by downloading executable code (optional) |

**REST API Benefits:**

- **Independent** — client and server are decoupled; changing the UI doesn't affect the API
- **Scalability** — the client/server separation allows each side to scale independently
- **Easy Integration** — ubiquitous HTTP protocol and support across many languages makes integration straightforward
- **Flexibility** — format-agnostic (JSON, XML, HTML); easy to understand and modify

### REST API Design Principles

**1. Accept & respond with JSON**

JavaScript has built-in methods to encode and decode JSON. Almost every networked technology supports it, making it the standard response format for REST APIs.

**2. Use nouns instead of verbs in endpoint paths**

The HTTP method already carries the verb — the path should name the resource:

```
❌  GET /getAllStudents
✅  GET /students
```

**3. Allow filtering, ordering, and pagination**

Limit the amount of data the API returns to conserve bandwidth:

```
GET /students?limit=10&offset=25
```

**4. Version the REST API**

Versioning communicates which features and resources are exposed. Use `/v1/`, `/v2/` etc. at the start of the path:

```
https://schooldirectory.com/v2/teachers
```

**5. Use caching**

Return data from a local cache instead of querying the database on every request. Middleware such as `redis` or `apicache` handles this:

```javascript
const app = express();
let cache = redisClient;
app.use(cache('5 minutes'));
```

**6. Proper HTTP error handling**

Return meaningful HTTP status codes so clients know what happened:

| Code | Meaning |
|---|---|
| `200 OK` | Success for GET, PUT, POST |
| `201 Created` | Resource created; body may be empty |
| `404 Not Found` | Requested resource unavailable |
| `500 Internal Server Error` | System-level failure |

```javascript
app.post('/students', (req, res) => {
  const { email } = req.body;
  const studentExists = users.find(u => u.email === email);
  if (studentExists) {
    return res.status(404).json({ error: 'Student already exists' });
  }
  // ...
});
```

**7. Document your API**

Start documentation at the design phase. [Swagger](https://swagger.io/) (built around the OpenAPI Spec) is a popular tool for generating interactive API documentation.

### Planning a REST API

A structured planning approach before writing code:

**Step 1 — Understand WHY you are building the API**
- Who is the target user?
- What products, services, and data should they access?
- What technologies need to integrate with the API?

**Step 2 — List out the user FUNCTIONALITY**
- What does the API need to do?
- Build user stories from this list
- Break the work into a backlog

**Step 3 — Determine the TYPE of API**
- REST, SOAP, RPC, GraphQL?

**Step 4 — Build an initial roadmap**
- Define the long-term strategy and direction — this can evolve over time

### Authentication Methods for a REST API

> It is recommended to secure REST API access using SSL & TLS.

There are four fundamental approaches to authentication:

| Method | Description |
|---|---|
| **HTTP Basic Auth** | Simplest approach — sends username & password alongside every API call |
| **API Key Auth** | Generates a unique secret key (long string) passed with every request |
| **OAuth** | Orchestrates approvals automatically between API owner and service; developers can self-authorize |
| **No Auth** | No credentials required — occasionally used for internal-only APIs; not recommended |

### Storing REST API Data

There are several ways to store data in a Node.js REST API:

**Databases (recommended for production):**

JavaScript supports both relational and NoSQL databases:

| Relational | NoSQL |
|---|---|
| MS SQL | MongoDB |
| PostgreSQL | Cassandra |
| MySQL | LevelDB |
| SQLite | Redis |
| Oracle | CouchDB |

Install a database driver with npm, then connect using a `createConnection` call:

```javascript
// database.js — MySQL example
const mysql = require('mysql');

const dbconnection = mysql.createConnection({
  host: "localhost",
  user: "user",
  password: "p@ssw0rd145",
  database: "schooldir",
});

dbconnection.connect();
```

Then import it in your routes:

```javascript
const dbconnection = require("../database");
```

**Arrays (suitable for development/demos):**

For simple demos or prototypes, store data directly in an in-memory array:

```javascript
const teachers = [
  "John Smith",
  "Deshan Jackson",
  "Susan Johnson",
  "Alexis Frederickson"
];
```

The array data is passed back through REST endpoints but resets each time the server restarts. Use a real database for persistence in production.

**Populate `index.js`** with middleware, data, and routes:

```javascript
const express = require('express');
const bodyParser = require('body-parser');
const cors = require('cors');

const app = express();
const port = 3000;

// In-memory data store
const jokes = [
  { id: 1, joke: "Why don't scientists trust atoms? Because they make up everything." },
  { id: 2, joke: "I told a joke about construction... I'm still working on it." }
];

// Middleware
app.use(bodyParser.json());
app.use(cors());

// Root endpoint
app.get('/', (req, res) => {
  res.json({ message: 'Jokes REST API' });
});

// GET all items
app.get('/jokes', (req, res) => {
  res.json(jokes);
});

// GET single item by id
app.get('/jokes/:id', (req, res) => {
  const joke = jokes.find(j => j.id === parseInt(req.params.id));
  if (!joke) return res.status(404).json({ error: 'Not found' });
  res.json(joke);
});

// POST — create item
app.post('/jokes', (req, res) => {
  const newJoke = { id: jokes.length + 1, joke: req.body.joke };
  jokes.push(newJoke);
  res.status(201).json(newJoke);
});

// DELETE — remove item
app.delete('/jokes/:id', (req, res) => {
  const index = jokes.findIndex(j => j.id === parseInt(req.params.id));
  if (index === -1) return res.status(404).json({ error: 'Not found' });
  jokes.splice(index, 1);
  res.status(200).json({ message: 'Deleted' });
});

app.listen(port, () => console.log(`Hello world app listening on port ${port}!`));
```

Run without nodemon:

```bash
node index.js
```

### Testing with Postman

Postman is a GUI tool for building, testing, and modifying APIs — used by over 20 million developers. It has a free tier and works on Windows, macOS, and Linux.

**Capabilities:**
- Make HTTP requests: `GET`, `POST`, `PUT`, `PATCH`, `DELETE`
- Save and manage environments for later reuse
- Convert API calls to code (JavaScript, Python, etc.)
- Organise requests into collections

**Installing Postman:**

| Platform | Command |
|---|---|
| Windows | Download `.exe` from postman.com and run the installer |
| macOS | `brew install --cask postman` |
| Linux (Ubuntu) | `sudo snap install postman` |

### Testing with curl

curl is a command-line tool for making HTTP requests, often pre-installed on Linux-based systems. It supports GET, POST, PUT, PATCH, user authentication, FTP, cookies, Telnet, LDAP, and more.

**Common curl examples:**

```bash
# GET — fetch all resources
curl http://localhost:3000/jokes

# GET — fetch from a public API
curl --request GET --url https://moviesdb.com/movies/all

# POST — send form data
curl --data "name=John&surname=Doe" http://localhost:3000/jokes

# POST — send JSON
curl -X POST \
  -H "Content-Type: application/json" \
  -d '{"joke": "Why did the bicycle fall over? Because it was two-tired."}' \
  http://localhost:3000/jokes

# DELETE
curl -X DELETE http://localhost:3000/jokes/1
```

curl is particularly useful in CI pipelines and scripts where a GUI is unavailable.
