## Module 01: Getting Started with JavaScript and ECMAScript 2025

_Introduces the JavaScript language, its runtimes and execution environments, version history, tooling (Node.js, npm, VS Code), and key online resources. Sets the stage for the entire course._

### What is JavaScript?

JavaScript is a **dynamic, interpreted programming language** that can be used to create web applications, back-end services, games, mobile applications, and just about anything else you can imagine.

**Where JavaScript runs:**
- Browsers
- Servers (Node.js)
- Command-line tools
- Desktop apps (Electron)
- Mobile apps
- Cloud and automation

### ECMAScript vs. JavaScript

The two terms are related but distinct:

````python
render_diagram(Path("assets/01-ecmascript-vs-javascript.png"))
````

- **ECMAScript** — the language standard, defined by Ecma TC39.
- **JavaScript** — the language as implemented in browsers, Node.js, and other environments.

ECMAScript defines what the language must do; JavaScript is each environment's concrete implementation of that specification.

### Engines and Runtimes

````python
render_diagram(Path("assets/02-engines-and-runtimes.png"))
````

- **Engine** — interprets, compiles, and executes JavaScript code (e.g. V8, SpiderMonkey, JavaScriptCore).
- **Runtime** — an engine paired with additional APIs that work in a specific environment (e.g. Node.js adds file I/O, networking; browsers add DOM, fetch).

This course uses Node.js 24 LTS, which runs on the V8 engine.

### JavaScript Versions

ECMAScript releases new capabilities annually. This course covers **up to ECMAScript 2025**, including:

- JSON modules
- Modern `Promise` helpers (`Promise.try`, `Promise.withResolvers`)
- Non-mutating collection patterns (`.toSorted()`, `.toSpliced()`, `.with()`)

> **Compatibility note:** The oldest JavaScript syntax still runs in modern engines — nothing is ever removed. Older patterns you may encounter: `var`, older loop styles, CommonJS `require()`, callback-based APIs.

## Module 02: Syntax, Variables, and Primitive Data Types

_Covers the basic grammar of JavaScript: statements, expressions, keywords, identifiers, variable declarations, and the seven primitive types._

### Syntax

**Syntax** is the set of rules that define how characters and symbols must be structured in the language.

#### Statements

A statement is a complete instruction. Each statement typically ends with a semicolon (though they are optional due to Automatic Semicolon Insertion):

```js
console.log("Hello, JavaScript");
let firstName = "Andrey";
let lastName = "Stoev";
console.log(firstName);
```

#### Expressions

An expression produces a value and can appear inside a larger statement:

```js
2 + 3           // numeric expression
"Hello"         // string expression
let total = 2 + 3;  // assignment using an expression
```

#### Keywords and Identifiers

```js
let firstName = "Andrey";
^^^                        // keyword (reserved word)
    ^^^^^^^^^              // identifier (name you choose)
```

**Naming rules:** identifiers must start with a letter, `_`, or `$`; they are case-sensitive; use camelCase by convention.

### Variables

**Best practice:** always use descriptive names.

```js
// Bad
let x = "Andrey";

// Good
let firstName = "Andrey";
let lastName = "Stoev";
let numberOfYearsEmployed = 6;
```

#### Declarations: `let`, `const`, `var`

```js
// let — re-assignable, block-scoped
let numberOfYearsEmployed = 6;
numberOfYearsEmployed = 7;           // OK

// const — not re-assignable, block-scoped (preferred default)
const employeeId = "E-1001";
// employeeId = "E-1002";           // TypeError

// var — function-scoped, hoisted (legacy; avoid in modern code)
var department = "Engineering";
```

#### Dynamic Typing

Variables hold a type at runtime, not compile time. The type of a variable can change:

```js
let value = "Andrey";
value = 42;
value = true;
```

### Value Categories

JavaScript has two categories of values: **primitive values** and **objects**.

#### Primitive Types

| Type | Example |
|---|---|
| `String` | `"Andrey"` |
| `Number` | `6`, `42.5` |
| `BigInt` | `9007199254740993n` |
| `Boolean` | `true`, `false` |
| `Symbol` | `Symbol("id")` |
| `Undefined` | `let manager;` (no assignment) |
| `Null` | `let manager = null;` (intentionally empty) |

```js
let firstName = "Andrey";
let numberOfYearsEmployed = 6;
let hourlyRate = 42.5;
let hugeNumber = 9007199254740993n;
let isActive = true;
let id = Symbol("id");
let manager;          // undefined
let supervisor = null;
```

Primitive values are **immutable** — you cannot change the value itself, only reassign the variable. They do have wrapper objects (`String`, `Number`, etc.) that provide methods:

```js
let name = "Andrey";
let uppercaseName = name.toUpperCase();  // "DAVID"
```

#### Object Types

Objects are mutable collections of named properties and the basis for all non-primitive values:

```js
// Object literal
let employee = { firstName: "Andrey", lastName: "Stoev" };
employee.lastName = "Smith";  // mutation is allowed

// Array
let departments = ["Engineering", "Marketing", "HR"];

// Map and Set
let employeesById = new Map();
let uniqueDepartments = new Set();

// Function
function sayHello() { console.log("Hello"); }

// Built-in object types
let startDate = new Date();
let error = new Error("Something went wrong");
let result = Promise.resolve("Done");
let pattern = /JavaScript/;
let bytes = new Uint8Array([1, 2, 3]);
```

Use `typeof` to check primitive types at runtime:

```js
console.log(typeof "Andrey");      // "string"
console.log(typeof 42);           // "number"
console.log(typeof true);         // "boolean"
console.log(typeof undefined);    // "undefined"
console.log(typeof null);         // "object"  ← historic quirk
console.log(typeof {});           // "object"
console.log(typeof []);           // "object"
```

## Module 03: Objects, Type Conversions, and Modern Syntax

_Deep dive into JavaScript objects, JSON, dates, null/undefined distinctions, classes, type coercion, and the modern destructuring/spread/chaining syntax patterns._

### JavaScript Objects

An object is a collection of **key-value pairs** (properties). Access properties with dot notation or bracket notation:

```js
const employee = {
  firstName: "Andrey",
  lastName: "Stoev",
  department: "Engineering",
  greet() {
    console.log(`Hello, I'm ${this.firstName}`);
  }
};

console.log(employee.firstName);        // dot notation
console.log(employee["lastName"]);      // bracket notation
employee.greet();
```

**JSON vs. object literals:** JSON is a text format based on JS object syntax but stricter — property names must be double-quoted strings, and function values are not allowed.

```json
{
  "firstName": "Andrey",
  "lastName": "Stoev"
}
```

### Dates and Times

Working with dates involves tricky complexity:
- Time zones and daylight saving time
- Different calendar systems
- Different regional formats
- Different precision requirements

JavaScript's built-in `Date` object represents a **single point in time** (milliseconds since Unix epoch). It is not a calendar. For display, always use formatting functions:

```js
// Creating dates
const now = new Date();
const specific = new Date("2025-01-15");

// Formatting with Intl (modern, locale-aware)
const formatter = new Intl.DateTimeFormat("en-US", { dateStyle: "long" });
console.log(formatter.format(now));  // "October 21, 2026"
```

### Null, Undefined, and Nullish Values

Two built-in "empty" values with different semantic meanings:

| Value | Meaning |
|---|---|
| `undefined` | Variable declared but no value assigned |
| `null` | Intentionally set to "no value" |

**Nullish values:** `null` and `undefined`

**Falsy but not nullish:**

```js
// Nullish values
null
undefined

// Falsy but not nullish
0
""
false
NaN
0n
```

**Nullish coalescing operator** (`??`) — returns the right side only if the left is `null` or `undefined`:

```js
const name = employee.firstName ?? "Unknown";
```

### Classes

JavaScript classes provide syntactic sugar over prototype-based inheritance:

```js
class Employee {
  constructor(firstName, lastName, department) {
    this.firstName = firstName;
    this.lastName = lastName;
    this.department = department;
    this.isActive = true;
  }

  getDisplayName() {
    return `${this.firstName} ${this.lastName}`;
  }
}

const emp = new Employee("Andrey", "Stoev", "Engineering");
console.log(emp.getDisplayName());  // "Andrey Stoev"
```

### Type Conversions

JavaScript performs **implicit coercion** in many contexts. It also supports explicit conversion:

```js
// Explicit conversion
Number("42")         // 42
String(42)           // "42"
Boolean(0)           // false
Boolean("Andrey")     // true

// Implicit coercion (careful!)
"5" + 3              // "53"  (+ triggers string concatenation)
"5" - 3              // 2     (- forces numeric)
```

**Falsy values** (evaluate to `false` in boolean context):

```js
false, 0, "", null, undefined, NaN, 0n
```

All other values are **truthy**, including empty objects `{}` and empty arrays `[]`.

### Modern Syntax Patterns

ECMAScript 2015+ introduced ergonomic syntax patterns widely used today:

```js
const employee = { firstName: "Andrey", lastName: "Stoev", department: "Engineering" };

// Destructuring
const { firstName, lastName } = employee;
const [first, second] = departments;

// Default values in destructuring
const { firstName, role = "Employee" } = employee;

// Spread syntax
const copy = { ...employee };
const merged = { ...employee, department: "Product" };
const combined = [...arr1, ...arr2];

// Rest syntax (in function params)
function log(first, ...rest) { console.log(rest); }

// Optional chaining
const city = employee.address?.city;  // undefined if address is absent

// Nullish coalescing
const displayName = employee.preferredName ?? employee.firstName;

// Template literals
const greeting = `Hello, ${firstName} ${lastName}!`;
```

### Regular Expressions

Regular expressions (regex) enable pattern matching in text:

```js
const pattern = /JavaScript/i;         // literal syntax, case-insensitive
const re = new RegExp("JavaScript", "i");  // constructor form

// Test for a match
pattern.test("I love JavaScript");     // true

// Find matches
"JavaScript is great".match(/\w+/g);  // ["JavaScript", "is", "great"]

// Replace
"Hello World".replace(/World/, "JS"); // "Hello JS"

// Modern: String.prototype.matchAll
for (const match of "aababc".matchAll(/a(b*)/g)) {
  console.log(match[0], match[1]);
}
```

## Module 04: Expressions, Operators, and Conditional Logic

_Covers mathematical, assignment, comparison, and logical operators, plus truthy/falsy evaluation and `if`/`else` conditional logic._

### Expressions

An **expression** is any code that produces a value:

```js
2 + 3                           // numeric expression → 5
"Engineering"                   // string expression
const department = "Engineering"; // assignment expression
```

### Mathematical Operators

```js
let total = 10 + 3;    // 13   addition
let diff  = 10 - 3;    // 7    subtraction
let prod  = 10 * 3;    // 30   multiplication
let quot  = 10 / 3;    // 3.33 division
let rem   = 10 % 3;    // 1    modulo (remainder)
let exp   = 2 ** 10;   // 1024 exponentiation
let inc   = 5; inc++;  // 6    post-increment
```

### Assignment Operators

```js
let x = 10;
x += 5;   // x = 15
x -= 3;   // x = 12
x *= 2;   // x = 24
x /= 4;   // x = 6
x **= 2;  // x = 36
x %= 5;   // x = 1
```

### Comparison Operators

```js
5 == "5"    // true  — loose equality (type coercion)
5 === "5"   // false — strict equality (no coercion)
5 !== "5"   // true
5 > 3       // true
5 < 3       // false
5 >= 5      // true
5 <= 4      // false
```

> **Rule:** use strict equality (`===` / `!==`) unless you have a specific reason not to. Loose equality has surprising results due to coercion.

### Logical Operators and Truthy/Falsy

```js
true && false    // false  (AND — both must be truthy)
true || false    // true   (OR  — at least one must be truthy)
!true            // false  (NOT)

// Short-circuit evaluation
const name = user && user.firstName;   // "Andrey" or falsy
const role = user.role || "Employee";  // first truthy wins
```

### Conditional Logic

```js
if (employee.isActive) {
  console.log("Active");
} else if (employee.department === "Engineering") {
  console.log("Engineer");
} else {
  console.log("Other");
}
```

## Module 05: Branching, Loops, and Input Validation

_Covers ternary operators, switch statements, all loop types (for, while, do-while, for-of, for-in), loop control flow, and combining these concepts for input validation._

### Ternary Operator

A compact conditional expression:

```js
const badge = yearsOfService >= 5 ? "yellow" : "blue";
// equivalent to:
// let badge;
// if (yearsOfService >= 5) { badge = "yellow"; } else { badge = "blue"; }
```

### Switch Statements

Best when choosing among many discrete values:

```js
switch (department) {
  case "Engineering":
    console.log("Tech team");
    break;
  case "Marketing":
    console.log("Marketing team");
    break;
  default:
    console.log("Other team");
}
```

### Loop Types

#### `for` — when you need an index or precise control

```js
for (let index = 0; index < 5; index++) {
  console.log(index);
}
```

#### `while` — when you don't know the count ahead of time

```js
let attempts = 0;
while (attempts < 3) {
  console.log("Trying again...");
  attempts++;
}
```

#### `do...while` — executes body at least once

```js
let attempts = 0;
do {
  console.log("Trying at least once...");
  attempts++;
} while (attempts < 3);
```

#### `for...of` — iterate over any iterable (arrays, strings, Maps, Sets)

```js
const departments = ["Engineering", "Marketing", "HR"];
for (const department of departments) {
  console.log(department);
}
```

#### `for...in` — iterate over object keys

```js
const employee = { firstName: "Andrey", lastName: "Stoev", department: "Engineering" };
for (const key in employee) {
  console.log(`${key}: ${employee[key]}`);
}
```

### Loop Control Flow

```js
for (const emp of employees) {
  if (emp.department === "skip") continue;  // skip this iteration
  if (someCondition) break;                 // exit the loop
}
```

## Module 06: Arrays, Objects, Maps, and Sets

_Covers the four primary JavaScript collection types, their use cases, and ECMAScript 2025's new `Set` operation methods._

### Collection Types

#### Array — ordered, indexed, allows duplicates

```js
const departments = ["Engineering", "Marketing", "HR"];
const employees = [employeeOne, employeeTwo];
departments[0];           // "Engineering"
departments.length;       // 3
departments.push("Legal"); // add to end
departments.pop();         // remove from end
```

#### Object — unordered key-value pairs, string/Symbol keys

```js
const employee = {
  firstName: "Andrey",
  lastName: "Stoev",
  department: "Engineering",
  isActive: true
};
```

#### Map — key-value pairs, any key type, preserves insertion order

```js
const employeesById = new Map();
employeesById.set("E-1001", employee);
employeesById.get("E-1001");  // returns employee
employeesById.has("E-1001");  // true
employeesById.size;           // 1
```

**Use Map over Object when:** keys are computed, keys are not strings, insertion order matters, or you need `size`.

#### Set — unique values only, any type, preserves insertion order

```js
const uniqueDepts = new Set(["Engineering", "Marketing", "Engineering"]);
// → Set {"Engineering", "Marketing"}  (duplicate removed)
uniqueDepts.has("Marketing");  // true
uniqueDepts.size;              // 2
```

### Iterables and Iterators

````python
render_diagram(Path("assets/03-iterable-iterator-concepts.png"))
````

- **Iterable** — any value that JavaScript can loop over (arrays, strings, Maps, Sets, generators).
- **Iterator** — an object that knows how to produce the next value in a sequence (has a `next()` method returning `{ value, done }`).

Arrays, Maps, and Sets are all iterables, which means you can use `for...of`, spread syntax, and destructuring with them.

### ECMAScript 2025 Set Methods

New non-mutating set operations return a new `Set`:

```js
const a = new Set([1, 2, 3]);
const b = new Set([2, 3, 4]);

a.union(b);           // Set {1, 2, 3, 4}
a.intersection(b);    // Set {2, 3}
a.difference(b);      // Set {1}
a.symmetricDifference(b); // Set {1, 4}
a.isSubsetOf(b);      // false
a.isSupersetOf(b);    // false
a.isDisjointFrom(b);  // false
```

## Module 07: Grouping, Iteration, and Data Flow

_Covers grouping data with `Object.groupBy`/`Map.groupBy`, choosing the right iteration strategy, lazy iterators and iterator helpers, CLI argument parsing, and a preview of data storage options._

### Grouping Data

ECMAScript 2024+ introduces static grouping methods:

```js
const employees = [
  { name: "Andrey", department: "Engineering" },
  { name: "Mina",  department: "Product" },
  { name: "Esther",department: "Engineering" },
];

// Group into a plain object
const byDept = Object.groupBy(employees, (e) => e.department);
// { Engineering: [...], Product: [...] }

// Group into a Map (preserves any key type)
const byDeptMap = Map.groupBy(employees, (e) => e.department);
```

### Choosing the Right Loop

| Situation | Loop |
|---|---|
| Need an index or precise control | `for` |
| Values from an iterable (array, string, Map, Set) | `for...of` |
| Object property keys | `for...in` |
| Transform/filter an array | `map`, `filter`, `reduce` |
| Side effects on each item | `forEach` |

### Iterators and Iterator Helpers

An **iterator** is an object with a `next()` method returning `{ value, done }`. Iterators are useful for **lazy processing** — they compute only as much as you consume, rather than building a full intermediate array.

ECMAScript 2025 adds **iterator helpers** — chainable methods on `Iterator.prototype`:

```js
// Iterator.from converts any iterable to a lazy iterator
const result = Iterator.from(employees)
  .filter((e) => e.isActive)       // lazy filter
  .map((e) => e.department)        // lazy transform
  .take(3)                         // take first 3
  .toArray();                      // materialize

// Other helpers
.drop(n)             // skip first n
.flatMap(fn)         // flatten one level
.forEach(fn)         // side effects
.every(fn)           // boolean check
.some(fn)
.find(fn)
.reduce(fn, init)
```

> **When to use iterators vs. arrays:** use iterators when you're processing large sequences and only need a subset of results, or when building a pipeline where early termination is possible.

## Module 08: Function Fundamentals, Scope, and Closures

_Covers function declarations, expressions, arrow functions, scope rules, and closures._

### Functions

A **function** is a reusable block of code designed to perform a specific task. Naming: use descriptive verb-noun names.

```
calculateYearsOfService
formatEmployeeName
validateRequiredValue
loadEmployees
```

#### Parameters and Arguments

```js
function greet(firstName) {         // firstName is a parameter
  console.log(`Hello, ${firstName}`);
}
greet("Andrey");                     // "Andrey" is the argument
```

#### Return Values

```js
function getDisplayName(firstName, lastName) {
  return `${firstName} ${lastName}`;
}
const displayName = getDisplayName("Andrey", "Stoev"); // "Andrey Stoev"
```

#### Default Parameters

```js
function greet(firstName, greeting = "Hello") {
  console.log(`${greeting}, ${firstName}`);
}
greet("Andrey");           // "Hello, Andrey"
greet("Andrey", "Hi");     // "Hi, Andrey"
```

### Function Declarations vs. Expressions vs. Arrow Functions

```js
// Declaration — hoisted (can call before definition)
function isActiveEmployee(employee) {
  return employee.isActive;
}

// Expression — not hoisted, assigned to a variable
const formatEmployeeName = function(employee) {
  return `${employee.firstName} ${employee.lastName}`;
};

// Arrow function — concise, implicit return for single-expression bodies
const getBadgeColor = (yearsOfService) => {
  if (yearsOfService < 5) return "blue";
  if (yearsOfService < 10) return "yellow";
  return "silver";
};

// Arrow function with implicit return (expression body)
const double = (n) => n * 2;
```

> **Arrow functions and `this`:** arrow functions do **not** create their own `this` binding. They inherit `this` from the enclosing lexical scope. Use regular functions (or class methods) when you need a `this` that refers to the object.

### Scope

**Scope** determines where a variable can be accessed. Variables created inside a function are only available inside that function.

```js
function printName() {
  const firstName = "Andrey";
  console.log(firstName);          // works
}
// console.log(firstName);         // ReferenceError — not in scope
```

**Block scope** (`let`/`const`) limits a variable to the surrounding `{ }` block:

```js
if (true) {
  let blockVar = "inside";
  const alsoBlock = "inside";
}
// blockVar and alsoBlock are not accessible here
```

`var` is **function-scoped** (not block-scoped) and **hoisted** — one of the reasons it is avoided in modern code.

### Closures

A **closure** is a function that captures variables from its surrounding scope, even after that scope has finished executing:

```js
function createCounter(start = 0) {
  let count = start;              // captured by the returned function
  return function increment() {
    count++;
    return count;
  };
}

const counter = createCounter(10);
counter();   // 11
counter();   // 12
counter();   // 13
```

Closures are how JavaScript preserves state in callbacks, event handlers, and factory functions.

## Module 09: Functional Patterns, Recursion, and Lazy Processing

_Covers higher-order array methods (map, filter, reduce), higher-order functions, function composition, recursion, and combining named functions with iterator helpers._

### Array Iteration with Functions

The functional array methods take callback functions and return new arrays (they do not mutate the original):

```js
const employees = [
  { name: "Andrey",  isActive: true,  yearsOfService: 7  },
  { name: "Mina",   isActive: false, yearsOfService: 2  },
  { name: "Esther", isActive: true,  yearsOfService: 12 },
];

// forEach — side effect on each element (no return value)
employees.forEach((e) => console.log(e.name));

// map — transform each element → new array, same length
const names = employees.map((e) => e.name);
// ["Andrey", "Mina", "Esther"]

// filter — keep elements matching predicate → new array, shorter/same length
const active = employees.filter((e) => e.isActive);

// reduce — accumulate elements into a single value
const totalYears = employees.reduce((sum, e) => sum + e.yearsOfService, 0);
// 21

// find — first element matching predicate (or undefined)
const senior = employees.find((e) => e.yearsOfService > 10);

// findIndex — index of first match (or -1)
const idx = employees.findIndex((e) => e.name === "Mina");

// every — true if all match
const allActive = employees.every((e) => e.isActive);  // false

// some — true if at least one matches
const anyActive = employees.some((e) => e.isActive);   // true

// sort (toSorted — non-mutating, ES2023+)
const sorted = employees.toSorted((a, b) => a.name.localeCompare(b.name));
```

### Higher-Order Functions

A **higher-order function** is one that takes another function as an argument or returns a function. They enable reusable, composable logic:

```js
// Factory: returns a validator function
function createMinLengthValidator(min) {
  return function validate(value) {
    return value.length >= min;
  };
}
const validateName = createMinLengthValidator(2);
validateName("Al");      // true
validateName("A");       // false

// Reusable formatter
function createEmployeeFormatter(format) {
  return function formatEmployee(employee) {
    return format(employee);
  };
}
const summaryFormatter = createEmployeeFormatter(
  (e) => `${e.firstName} ${e.lastName} — ${e.department}`
);
```

### Recursion

A function that calls itself. Every recursive function needs a **base case** (stop condition) to avoid infinite recursion.

**When to use recursion:** tree structures, nested data, hierarchies — anything that is naturally self-similar.

```
Nested folders     Nested comments
Nested departments Tree structures
Menu hierarchies   Organization charts
```

```js
// Example: sum a nested tree of departments
function sumEmployees(node) {
  // Base case: leaf node
  if (!node.children || node.children.length === 0) {
    return node.employeeCount;
  }
  // Recursive case: sum children
  return node.employeeCount + node.children.reduce(
    (total, child) => total + sumEmployees(child), 0
  );
}
```

### Functions and Lazy Pipelines

Combining named functions with iterator helpers gives clean, readable, and lazy pipelines:

```js
import { Iterator } from "node:iterator";  // Node.js 22+

function isActive(employee) { return employee.isActive; }
function toDepartment(employee) { return employee.department; }

const uniqueActiveDepts = Iterator.from(employees)
  .filter(isActive)
  .map(toDepartment)
  .toArray()
  .filter((dept, i, arr) => arr.indexOf(dept) === i);  // unique
```

## Module 10: Problem Solving and Debugging JavaScript

_Covers problem decomposition, translating pseudocode to code, tracing program execution, and debugging with VS Code's built-in debugger and console tools._

### Problem-Solving Approach

1. **Understand the problem** — read carefully, identify inputs and expected outputs.
2. **Break it down** — decompose into small, named steps. Each step = one function.
3. **Write pseudocode** — describe the algorithm in plain language before coding.
4. **Implement incrementally** — write and test one step at a time.

**Example problem:** _Given a list of employees, return a list of active Engineering employees with at least five years of service, sorted by years of service descending._

```
Pseudocode:
1. Filter employees where isActive = true
2. Filter to Engineering department only
3. Filter to yearsOfService >= 5
4. Sort by yearsOfService descending
5. Return result
```

```js
// Implementation follows the pseudocode steps directly
const result = employees
  .filter((e) => e.isActive)
  .filter((e) => e.department === "Engineering")
  .filter((e) => e.yearsOfService >= 5)
  .toSorted((a, b) => b.yearsOfService - a.yearsOfService);
```

### Tracing Program Execution

Before running code, **predict the output**. Trace through the code manually:
- Track variable values line by line
- Follow function calls and returns
- Note the order of operations in expressions

After running, compare your prediction to the actual output. Discrepancies reveal misunderstandings.

> **Practice:** before running code, pause and predict the output, then run it. If the output matches your prediction, you have a complete mental model. If it does not, you found something to investigate.

### Console Debugging Tools

```js
console.log(value);           // print a value
console.log({ variable });    // shorthand to show name + value

// Table for arrays of objects

// Assertions (throw if false)

// Groups

console.groupEnd();

// Timing

```

### Debugging with VS Code

1. Set a **breakpoint** by clicking to the left of a line number.
2. Run with `F5` or "Start Debugging" — execution pauses at the breakpoint.
3. **Step Over** (`F10`) — execute current line, stay at same level.
4. **Step Into** (`F11`) — follow a function call inside.
5. **Step Out** (`Shift+F11`) — finish current function and return to caller.
6. Use the **Variables** panel to inspect local and global variables.
7. Add expressions to the **Watch** panel to track specific values.
8. Use the **Debug Console** to evaluate expressions at the paused point.

> A key workflow: run the program, pause where the bug occurs, inspect the actual values — then compare to what you expected.

## Module 12: Asynchronous JavaScript

_Covers synchronous vs. asynchronous execution, callbacks, Promises, async/await, and error handling with try/catch/finally._

### Synchronous vs. Asynchronous

**Synchronous** code executes line by line. Each statement completes before the next begins:

```js
console.log("One");
console.log("Two");
console.log("Three");
// One → Two → Three (always in order)
```

**Asynchronous** code allows other work to happen while waiting for a slow operation (network, file I/O, timers):

```js
// setTimeout schedules work to run after a delay
setTimeout(() => {
  console.log("This runs later.");
}, 1000);
console.log("This runs first.");
// "This runs first." → (1 second) → "This runs later."
```

### The Three Patterns

#### 1. Callbacks (original, legacy-leaning)

```js
readFile("employees.json", (error, data) => {
  if (error) { console.error(error); return; }
  console.log(data);
});
```

#### 2. Promises (ES2015)

```js
fetch("https://api.example.com/employees")
  .then((response) => response.json())
  .then((employees) => console.log(employees))
  .catch((error) => console.error(error));
```

#### 3. Async/Await (ES2017) — preferred for readability

```js
async function loadEmployees() {
  const response = await fetch("https://api.example.com/employees");
  const employees = await response.json();
  return employees;
}
```

### Error Handling

#### Creating and Throwing Errors

```js
// Create manually
const error = new Error("Something went wrong.");
console.log(error.message);

// Throw from a function
function validateEmployee(employee) {
  if (!employee.firstName) {
    throw new Error("First name is required.");
  }
  return true;
}
```

#### try / catch / finally

```js
// Synchronous errors
try {
  const data = JSON.parse("not valid JSON");
} catch (error) {
  console.error("Invalid JSON:", error.message);
}

// With finally — always runs regardless of success or failure
try {
  validateEmployee({ firstName: "Andrey" });
} catch (error) {
  console.error(error.message);
} finally {
  console.log("Validation attempt complete.");
}
```

#### Errors with Promises

```js
// Promise chain
someAsyncOperation()
  .then((result) => console.log(result))
  .catch((error) => console.error(error.message));

// async/await + try/catch
try {
  const result = await someAsyncOperation();
  console.log(result);
} catch (error) {
  console.error(error.message);
}
```

> **Rule:** always handle errors in async code. An unhandled promise rejection in Node.js will crash the process in modern versions.

## Module 13: Promises, HTTP Requests, and Event Handling

_Covers Promise combinators, making HTTP requests with fetch, concurrent vs. sequential async flows, event-driven programming, and Node.js EventEmitter._

### Promise Combinators

A **promise combinator** coordinates multiple promises. Use the right one for the use case:

| Combinator | Behavior |
|---|---|
| `Promise.all(arr)` | Resolves when all resolve; rejects immediately if any reject (fail-fast) |
| `Promise.allSettled(arr)` | Waits for all to settle; returns array of `{status, value/reason}` |
| `Promise.any(arr)` | Resolves with the first resolved; rejects only if all reject |
| `Promise.race(arr)` | Resolves/rejects with the first settled (either direction) |
| `Promise.try(fn)` | ES2025 — starts a promise flow from a function that may return or throw synchronously |
| `Promise.withResolvers()` | ES2024 — returns `{promise, resolve, reject}` for external resolution |

```js
// All must succeed — if one fails, the whole batch fails
const [employees, departments] = await Promise.all([
  loadEmployees(),
  loadDepartments(),
]);

// Need every outcome regardless
const results = await Promise.allSettled([
  loadEmployees(),
  loadConfig(),
]);
results.forEach((r) => {
  if (r.status === "fulfilled") console.log(r.value);
  else console.error(r.reason);
});
```

> **Sequential vs. concurrent:** always ask — does this step depend on the previous result? If yes, use sequential (`await` one by one). If no, concurrent (`Promise.all`) is faster.

```js
  throw new Error(`HTTP error: ${response.status}`);

// POST with JSON body

  method: "POST",
```

### Event-Driven Programming

**Event-driven programming** means the program responds to events as they happen rather than executing a linear script. Handlers are registered in advance and invoked when the matching event fires.

#### Browser Events

```
Click events    Keyboard events    Form events
Mouse events    Page load events
```

#### Node.js Events

```
Stream events    Process events
Server events    Custom event emitters
```

#### Event Handler Pattern

```js
// The basic pattern — same across browsers and Node.js
something.on("eventName", () => {
  console.log("The event happened.");

```


```js
// A function should have one clear job
function isActiveEmployee(employee) {
  return employee.isActive === true;
}
```

#### Avoid Duplication

> Duplication makes code harder to maintain. If the same validation rule appears in five places and the rule changes, you must update all five. If you miss one, you have a bug. Extract to a shared function.

### Reading Existing JavaScript Code

When approaching an unfamiliar codebase, follow this sequence:

#### 1. Start with `package.json`

```json
{
  "type": "module",
  "scripts": {
    "start": "node src/index.js",
    "test": "vitest",
    "lint": "eslint ."
  },
  "dependencies": {},
  "devDependencies": {}
}
```

Scripts tell you how to run, test, and lint the project.

#### 2. Follow Imports

```js
// Imports show what this file depends on
import { readJsonFile }        from "./files.js";
import { validateEmployee }    from "./validation.js";
import { formatEmployeeSummary } from "./formatting.js";
```

#### 3. Find Exported Functions

```js
// Exports define the module's public surface
export function validateEmployee(employee) {
  // ...
}
```

#### 4. Trace Data Flow

```js
// Read → filter → format — the data-flow story
const employees     = await readJsonFile("./data/employees.json");
const activeEmps    = employees.filter((e) => e.isActive);
const summaries     = activeEmps.map(formatEmployeeSummary);
```

#### 5. Identify Legacy Patterns

```js
// var — legacy variable declaration
var firstName = "Andrey";

// CommonJS require / module.exports
const fs = require("fs");
module.exports = { loadEmployees };

// Error-first callback — predates promises/async-await
readFile("employees.json", (error, data) => { /* ... */ });
```

#### 6. Use Project Clues

- Comments and documentation explain **intent**
- Tests show **expected behavior**
- Linting rules reveal **team conventions**

### Linting with ESLint

A **linter** reads source code and reports issues based on rules. It accepts syntactically valid code but flags potential mistakes before they become bugs.

**ESLint** is the most common JavaScript linter.

```bash
npm install --save-dev eslint
npx eslint src/
```

```js
// eslint.config.js
import js from "@eslint/js";

export default [
  js.configs.recommended,
  {
    rules: {
      "no-unused-vars": "error",
      "no-undef": "error",
    },
  },
];
```

### Formatting with Prettier

A **formatter** rewrites source code in a consistent style. **Prettier** is the most common JavaScript formatter.

```bash
npm install --save-dev prettier
npx prettier --write src/
```

> The point of a formatter is not to catch mistakes. It is to **end the conversation about style** — code reviews stop being about formatting nitpicks and start being about what actually changed.

Most teams configure VS Code to run ESLint and Prettier automatically on save (ESLint extension + Prettier extension, set Prettier as default formatter, enable format-on-save).

### Modern JavaScript Compatibility Checklist

When adopting a new JS feature:

1. **Check runtime support** — does Node.js 24 or your target browser support it?
2. **Understand standard vs. proposal stage** — TC39 proposals move through stages 0–4; only stage 4 is in the spec.
3. **Distinguish standard from runtime APIs** — `fetch` is in the spec; `node:fs` is Node.js-specific.
4. **Use documentation** — MDN, Node.js docs, compatibility tables.
5. **Recognize legacy patterns** — `var`, CommonJS, callbacks indicate older code.
