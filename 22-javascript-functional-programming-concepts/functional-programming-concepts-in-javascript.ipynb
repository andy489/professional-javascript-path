## Module 01: What Is Functional Programming? <a id="module-01"></a>

An introduction to the functional programming paradigm — what it is, how it compares to imperative and object-oriented programming, and why it matters for JavaScript developers.

### Definition

> Functional programming is a programming paradigm — a style of building the structure and elements of computer programs, that treats computation as the evaluation of mathematical functions and avoids change-state and mutable data.
>
> — Wikipedia

### Programming Paradigms

| Imperative | Declarative |
|---|---|
| Procedural and Object-oriented Programming | Functional, Mathematical, and Reactive Programming |
| Focuses on the **"How"** | Focuses on **"What"** |
| States the order in which operations occur | Does not state the order in which to execute operations |

**Imperative example** — explicitly iterates and mutates:

```javascript
for (var i = 0; i < courses.length; i++) {
  courses[i].lastModified = new Date()
}
```

**Declarative example** — describes the transformation:

```javascript
courses.map((course) => {
  course.lastModified = new Date();
  return course
})
```

### Key Principles

#### 1. Functions Do One Thing Well

In FP, each function should have a single, clear responsibility. This makes code readable, maintainable, and testable.

**Anti-pattern — one function doing too much:**

```javascript
function getCourseInformation() {
  // 1. Call API and get courses information.
  // 2. Sort courses based on release date, with newest being first.
  // 3. Render the courses within the UI.
}
```

**FP pattern — three focused functions:**

```javascript
function getCourseInfoQuery() {
  // Call API and get courses information.
}

function sortCourses() {
  // Sort courses based on release date, with newest being first.
}

function renderCourses() {
  // Render the courses within the UI.
}
```

Benefits: **Readable** · **Maintainable** · **Testable**

#### 2. Immutable Data

> Functional Programming is all about building functions for immutable variables.

Immutable data cannot be changed after it is created. Functions operate on copies, not in-place mutations.

```javascript
const initialArray = ['c', 'd', 'a', 'e', 'b'];
const sortedArray = sortFunction(initialArray);
// sortedArray = ['a', 'b', 'c', 'd', 'e']  ← new array
// initialArray = ['c', 'd', 'a', 'e', 'b'] ← unchanged
```

`sortFunction` does not replace `initialArray` — it returns a new `sortedArray`. Two distinct arrays exist after the call.

### Why Functional Programming?

- Code is **concise, predictable, and readable**
- **Modular code** with smaller, independent units
- **Easier to test** functional code
- Results in **fewer bugs**

### FP vs. Object-oriented Programming

| | Functional Programming | Object-oriented Programming |
|---|---|---|
| **Primary unit** | Function | Object |
| **Data** | Immutable | Mutable |
| **Data structures** | Fewer | More |
| **Side effects** | Functions have no side effects | Methods often have side effects |

### JavaScript and FP

JavaScript supports **both** object-oriented and functional programming paradigms. You can choose whichever paradigm suits your application's use case, or use a combination of both.

## Module 02: Explore Core Features of Functional Programming <a id="module-02"></a>

A deep dive into the building blocks of FP in JavaScript: first class functions, pure functions, side effects, and closures — with a real-world Redux example.

> FP is the process of building software by composing pure functions, avoiding shared state, mutable data, and side-effects.
>
> — Eric Elliott

### First Class Functions

In FP, **functions are treated as a data type** — a first-class citizen, just like numbers or strings.

JavaScript supports first class functions by default. They can be:

- **Assigned to variables**
- **Passed as arguments** to other functions
- **Returned as results** from other functions
- **Included in data structures** (arrays, objects)

Benefits: **Reusable** · **Flexible** · **Predictable** · **Testable** · **Readable** · **Maintainable**

````python
// Assign a function to a variable
const topic = (t) => `You are learning ${t}`;
console.log(topic("First class functions"));

// Pass a function as an argument
const numbers = [1, 2, 3, 4];
const multiplyTwo = (number) => number * 2;
const timesTwo = numbers.map(multiplyTwo);
console.log(timesTwo); // [2, 4, 6, 8]

// Return a function from another function
const firstClassFunction = () => () =>
  console.log("Hi! I am a function returned from another function");

let displayFunction = firstClassFunction();
displayFunction();
````

### Pure Functions

A **pure function** has two characteristics:

1. Given the same input, **always returns the same output**
2. **Produces no side-effects**

```javascript
const topic = (t) => `You are learning ${t}`;
// topic("closures") always returns "You are learning closures"
```

### Side Effects

A **side effect** is any interaction with the outside world — outside of the function's own scope.

**Common side effects:**

| Category | Examples |
|---|---|
| **Data mutation** | Changing or mutating the input |
| **DOM** | Querying or updating the DOM object |
| **Network** | Making API calls via `fetch`/`xhr` |
| **I/O** | Logging, reading/writing to a file |
| **Global state** | Reading from a global variable |

> A pure function should be **side-effect free!**

Side effects are unavoidable in real applications — the goal of FP is to **isolate** them, keeping the core logic pure and the side effects at the edges.

### Closures

> A closure gives you access to an **outer function's scope** from an **inner function**. In JavaScript, closures are created every time a function is created, at function creation time.
>
> — MDN Web Docs

Closures are a fundamental mechanism that enables many FP patterns — including data encapsulation, partial application, and memoization.

````python
// Outer function with enclosed variable
const welcomeAuthor = () => {
  let name = "Adhithi";
  const authorName = () => "Welcome " + name; // inner function closes over `name`
  return authorName;
};

const greet = welcomeAuthor();
console.log(greet()); // "Welcome Adhithi"

// Practical closure: encapsulated counter state
const updateClickCounter = () => {
  var counter = 0;
  const incrementCounter = () => counter++;
  return incrementCounter;
};

const counterResults = updateClickCounter();
console.log(counterResults()); // 0
console.log(counterResults()); // 1
console.log(counterResults()); // 2
// Each call increments the same enclosed `counter` variable
````

### Real-world Example: Redux

**Redux** is a widely-used state management library that applies FP principles directly.

#### Reducers Are Pure Functions

A Redux reducer takes the current state and an action, and returns a **new** state — it never mutates the existing state.

```
function(state, action) => newState
```

Think of reducers like coffee makers: coffee powder (state) + water (action) = fresh cup of coffee (new state).

```javascript
function courseDetails(state = initialState, action) {
  switch (action.type) {
    case RATE_COURSE:
      return {
        ...state,          // spread operator creates a new copy
        rating: action.rate,
      };
    default:
      return state;
  }
}
```

Key properties of this reducer:
- **Pure function** — same state + same action always produces the same new state
- **No side-effects** — never touches external systems
- **Immutable** — uses the spread operator (`...`) to return a new copy rather than mutating `state`

## Module 03: Function Composition <a id="module-03"></a>

Function composition is the technique of combining multiple functions into a pipeline — the output of one becomes the input of the next.

````python
render_diagram(Path("assets/01-function-composition-flow.png"))
````

The diagram shows how **combining functions is function composition** — data flows through Func A → Func B → Func C, each function's output feeding the next.

In JavaScript, the basic pattern is nesting:

```javascript
function1(function2(value))
```

### Why Function Composition?

- **Less code** — eliminates intermediate variables
- **Improves readability** — intention is expressed as a chain of transformations
- **Easy to modify** — adding or removing a step is like adding/removing a link in a chain
- **Used by frameworks** — React uses composition by wrapping components around each other

### Ramda Library

Ramda is a JavaScript library purpose-built for functional programming.

| Feature | Description |
|---|---|
| **Pure functional style** | All functions are pure and side-effect free |
| **Immutability** | Functions never mutate their arguments |
| **Clean API** | Elegant, consistent interface |

**Installation:**

```bash
npm install ramda
```

```javascript
const R = require('ramda');
```

### `pipe` — Left-to-right Composition

`pipe` applies functions **left to right**: the first function receives the initial value, and each subsequent function receives the previous result.

```javascript
const { pipe } = require('ramda');

const triple = x => x * 3;
const cube   = x => x * x * x;

const output = pipe(cube, triple)(2);
// cube(2) = 8, then triple(8) = 24
console.log(output); // 24
```

### `compose` — Right-to-left Composition

`compose` applies functions **right to left** — mathematical convention where the rightmost function runs first.

```javascript
const { compose } = require('ramda');

const triple = x => x * 3;
const cube   = x => x * x * x;

const output = compose(cube, triple)(2);
// triple(2) = 6, then cube(6) = 216
console.log(output); // 216
```

**Rule of thumb:** use `pipe` when reading left-to-right feels natural for your pipeline; use `compose` when you want mathematical function notation `f(g(x))`.

## Module 04: Built-in JavaScript Functions <a id="module-04"></a>

JavaScript ships with several built-in array methods that embody the functional programming mindset. The FP approach: **don't worry about _how_ the problem is solved — focus on _what_ the problem is**.

The four core functional array methods: **`map`** · **`filter`** · **`flat`** · **`reduce`**

### `map`

`map()` creates a **new array** populated with the results of calling a provided function on every element in the calling array. The original array is unchanged.

```
[1, 2, 3, 4, 5]  .map(addTwo)  →  [3, 4, 5, 6, 7]
```

````python
const initialNumbers = [1, 2, 3, 4, 5];

const addTwo = (number) => number + 2;
const newNumbers = initialNumbers.map(addTwo);

console.log("Initial Numbers:", initialNumbers); // [1, 2, 3, 4, 5] — unchanged
console.log("New Numbers:", newNumbers);         // [3, 4, 5, 6, 7]
````

### `filter`

`filter()` creates a **shallow copy** of a portion of a given array, containing only the elements that pass the callback function's test. Elements that do not pass are excluded from the new array.

```
[1, 2, 3, 4, 5]  .filter(isOdd)  →  [1, 3, 5]
```

````python
const words = [
  "Courses",
  "Rating",
  "JavaScript",
  "Functional Programming",
];

const isWordGreaterThanSeven = (word) => word.length > 7;
const result = words.filter(isWordGreaterThanSeven);

console.log(result); // ["JavaScript", "Functional Programming"]
````

### `flat`

`flat()` creates a **new array** with all sub-array elements concatenated into it, recursively up to the specified depth. `flat()` is itself implemented recursively.

```
[1, 2, 3, [4, 5]]  .flat()  →  [1, 2, 3, 4, 5]
```

````python
const arr1 = [1, 2, 3, 4, [5, 6]];
console.log(arr1.flat()); // [1, 2, 3, 4, 5, 6]  — default depth 1

const arr2 = [1, 2, 3, 4, [[[5, 6]]]];
console.log(arr2.flat(2)); // [1, 2, 3, 4, [5, 6]]  — depth 2, one level remains
````

### `reduce`

`reduce()` is a **central concept in FP**. It:

1. Runs a reducer callback function over all elements in the array
2. Runs in ascending-index order, accumulating results
3. Reduces an array down to a **single value**

```
[1, 2, 3, 4, 5]  .reduce(multiply)  →  120
```

`reduce()` does not mutate the array it is called on, though the provided callback function can.

````python
const initialArray = [1, 2, 3, 4, 5];

// Multiply all elements: 1 * 2 * 3 * 4 * 5
const multiply = initialArray.reduce(
  (accumulator, currentValue) => accumulator * currentValue
);
console.log("Multiply result:", multiply); // 120

// Find the maximum value
const getMax = (a, b) => Math.max(a, b);
const max = initialArray.reduce(getMax);
console.log("Max result:", max); // 5

// Edge case: reduce skips missing elements (sparse array) but NOT undefined
console.log([1, 2, , 4].reduce((a, b) => a + b));        // 7 (hole skipped)
console.log([1, 2, undefined, 4].reduce((a, b) => a + b)); // NaN (undefined included)
````

**Note on `reduce` readability:** `reduce` can be difficult to understand at a glance. Developers should weigh the readability tradeoffs against its power, and add clear documentation when using it in shared code.

## Module 05: Advanced Concepts of Functional Programming <a id="module-05"></a>

Three powerful FP techniques: recursion, higher-order functions, and currying.

### Recursion

**Recursion** is a technique where a function calls itself. It needs a **stopping condition** (base case), or it will call itself forever.

Recursive functions let you perform a unit of work multiple times — they are the FP alternative to imperative loops. `flat()` is a built-in example of a recursive function.

Key points:
- Needs a base case — a stopping point
- Can be harder to reason about than loops, but expresses the problem more declaratively

````python
// Imperative (loop) approach
const incrementByOne = (number) => {
  for (let i = number; i < 11; i++) {
    console.log(i);
  }
};

// Recursive (FP) approach — same output, no mutable loop variable
const incrementByOneRecursive = (number) => {
  if (number === 11) return; // base case: stop condition
  console.log(number);
  incrementByOneRecursive(number + 1); // recursive call with reduced problem
};

incrementByOneRecursive(1); // prints 1 through 10
````

### Higher-order Functions

A **higher-order function** (HOF) is a function that:

1. **Accepts a function as an argument**, and/or
2. **Returns a function**

The built-in array methods `map()`, `filter()`, and `reduce()` are all higher-order functions — they all accept a callback function as an argument.

````python
// Built-in HOFs
const initialNumbers = [1, 2, 3, 4, 5];
const addTwo = (number) => number + 2;
const newNumbers = initialNumbers.map(addTwo); // map is a HOF

const isWordGreaterThanSeven = (word) => word.length > 7;
const words = ["JavaScript", "Courses", "Functional Programming"];
const longWords = words.filter(isWordGreaterThanSeven); // filter is a HOF

const getMax = (a, b) => Math.max(a, b);
const max = initialNumbers.reduce(getMax); // reduce is a HOF
console.log(max); // 5

// Custom higher-order function
const add      = (n1, n2) => n1 + n2;
const subtract = (n1, n2) => n1 - n2;
const multiply = (n1, n2) => n1 * n2;

// `compute` is a HOF: it accepts a math operation function as its first argument
const compute = (mathOperation, initialValue, values) => {
  let total = initialValue;
  values.forEach((value) => {
    total = mathOperation(total, value);
  });
  return total;
};

console.log(compute(add,      0,  [2, 4])); // 6
console.log(compute(subtract, 10, [5, 2])); // 3
console.log(compute(multiply, 1,  [2, 4])); // 8
````

### Currying

**Currying** transforms a function with multiple arguments into a chain of functions, each containing a single argument. It wraps a function inside a function.

This is achieved naturally in JavaScript through closures — each returned function captures the argument from its enclosing scope.

````python
// Curried function: 3 arguments become 3 chained single-argument functions
const sendMessage = (greeting) => (name) => (message) =>
  `${greeting} ${name} ${message}`;

console.log(sendMessage("Hello")("Adhithi")("You are learning Currying"));
// "Hello Adhithi You are learning Currying"

// Curried addition
const add = (a) => (b) => (c) => a + b + c;
console.log(add(5)(10)(20)); // 35

// Partial application: supply arguments one at a time
const helloGreeting = sendMessage("Hello");      // returns (name) => ...
const helloToAdhithi = helloGreeting("Adhithi"); // returns (message) => ...
console.log(helloToAdhithi("Currying is powerful!"));
````

## Module 06: JavaScript Frameworks and FP <a id="module-06"></a>

How modern JavaScript frameworks — React and Svelte — apply or relate to functional programming principles.

### React Basics

React is a JavaScript library for building rich user interfaces. UIs are built from small units — buttons, text, images — combined into reusable, nestable **components**.

Each component is a building block: a reusable piece of UI. Assembling them results in a complete application.

````python
render_diagram(Path("assets/02-react-component-hierarchy.png"))
````

The component tree illustrates the **component-based model**: data flows one-way from parent to child. `App` is the root; `NavBar`, `SearchBar`, `Suggestion`, and `Feed` are its children; `Follow`, `Tweet`, and `Like` nest further down.

### Functional Programming in React

React assumes that **every component you write is a pure function**.

A **pure React component** returns the same JSX given the same inputs (props):

```jsx
function CoffeeRecipe({ guests }) {
  return (
    <ol>
      <li>Boil {guests} cups of water.</li>
      <li>Add {guests} spoons of coffee.</li>
      <li>Add {0.5 * guests} cups of milk to boil and sugar to taste.</li>
    </ol>
  );
}
```

**Props are read-only** — a component must never modify its own props.

#### Avoiding Side Effects in Components

Side effects inside render are a violation of the pure component contract:

| | Impure ❌ | Pure ✓ |
|---|---|---|
| **Mutation** | `guests = guests + 1` outside component | Receives `guests` as a prop |
| **Result** | Different output each call | Same JSX for same props |

#### Where Do Side Effects Go?

Change within an app is inevitable. **Event handlers** are the designated place for side effects in React — functions that React runs when you perform an action (click, submit, etc.). Event handlers do not have to be pure.

#### Benefits of Keeping Components Pure

- **Server rendering** — pure components can serve many user requests from the same instance
- **Performance** — React can skip re-rendering components whose inputs haven't changed; pure functions are safe to cache
- **Unlocks the React paradigm** — purity is the foundation that makes React's reconciliation model work correctly

Reference: [react.dev/learn/keeping-components-pure](https://react.dev/learn/keeping-components-pure)

### Svelte

Svelte takes a different approach to building user interfaces — it is an alternative to React and Vue that shifts work from the browser to the compiler.

> Reactive programming is basically a program that reacts to events over time. Svelte lets you write reactive code with JavaScript!

**Key differences from React:**

| Aspect | React | Svelte |
|---|---|---|
| **Components** | Must be pure functions | Flexible — do not have to be pure |
| **Paradigm** | Functional programming | Reactive programming |
| **Work location** | Runtime (browser) | Compile-time (compiler) |
| **FP opt-in** | Required (purity enforced) | Optional (`immutable` option available) |

#### Svelte's Immutable Option

You can opt into immutability per component when you want FP guarantees:

```svelte
<svelte:options immutable={true} />;
```

This tells the Svelte compiler to expect immutable data for that component — enabling it to use faster change-detection. By default, `immutable` is `false`.

Svelte allows you to use functional programming concepts like **immutable data** and **pure functions** if you choose to, without mandating them.
