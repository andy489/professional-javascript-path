## Module 01: Installing an IDE

An IDE (Integrated Development Environment) is a text editor with extra tooling that makes working on projects more efficient. This module covers choosing, installing, and configuring Visual Studio Code as the primary IDE for JavaScript development.

### What Is an IDE?

An IDE provides features beyond basic text editing:

- **Syntax highlighting** — color-codes language constructs for readability
- **Code completion / IntelliSense** — suggests completions as you type
- **Error detection** — flags syntax and type errors in real time without running the code
- **Integrated terminal** — run shell commands without leaving the editor
- **Debugger** — step through code, inspect variables, set breakpoints
- **Extensions / plugins** — add language support, linters, formatters, and more
- **Source control integration** — diff, commit, and branch from within the editor

### Choosing an IDE for JavaScript

Several popular options exist:

| IDE | Notes |
|---|---|
| **Visual Studio Code** | Free, open-source, cross-platform, huge extension ecosystem — the most widely used JavaScript IDE |
| **WebStorm** | Commercial, deep JavaScript/TypeScript support out of the box |
| **Vim / Neovim** | Terminal-based, highly customisable, steep learning curve |
| **Sublime Text** | Lightweight, fast, commercial licence |

This course uses **Visual Studio Code** (VS Code).

### Installing Visual Studio Code

Download from [code.visualstudio.com](https://code.visualstudio.com). Installers are available for Windows, macOS, and Linux.

After installation, launch VS Code from the Applications folder (macOS), Start menu (Windows), or your desktop environment (Linux).

### Configuring VS Code

#### User Settings vs. Workspace Settings

VS Code has two levels of settings:

- **User settings** — apply globally to every project you open (stored in your OS profile)
- **Workspace settings** — stored in `.vscode/settings.json` inside the project folder; override user settings and are committed with the project

To open settings: `Ctrl+,` (Windows/Linux) or `Cmd+,` (macOS), or use the Command Palette (`Ctrl+Shift+P` / `Cmd+Shift+P`) and search for "Open Settings".

#### Useful Settings to Configure

| Setting | What it does |
|---|---|
| `editor.formatOnSave` | Automatically formats the file when you save |
| `editor.defaultFormatter` | Chooses which formatter runs on save (e.g., Prettier) |
| `editor.tabSize` | Number of spaces per indentation level |
| `editor.wordWrap` | Wraps long lines |
| `files.autoSave` | Saves files automatically |

Example `.vscode/settings.json` for a JavaScript project:

```json
{
  "editor.formatOnSave": true
}
```

The **Workspace** tab in Settings UI sets this just for the current project folder, while **User** sets it globally.

#### Selecting a Formatter

When `editor.formatOnSave` is enabled, VS Code may ask which formatter to use. **Prettier** is the most widely adopted JavaScript formatter. Install it from the Extensions panel (`Ctrl+Shift+X` / `Cmd+Shift+X`) by searching for "Prettier - Code formatter".

Once installed, it will appear in the formatter picker as **Prettier - Code formatter (default)**.

### Troubleshooting a VS Code Installation

Before installing, verify your system is ready:

- **Sufficient disk space** — VS Code itself is small but extensions and projects add up
- **Antivirus software** — may need to whitelist VS Code's update process
- **Network connectivity** — needed for downloading extensions
- **Firewall rules** — VS Code connects to several Microsoft hostnames for updates, extensions, and telemetry

#### Common Issues and Fixes

**Set up a network connection:**  
VS Code requires network access for its update mechanism and the Extension Marketplace. See `https://code.visualstudio.com/docs/setup/network` for required hostnames to whitelist.

**Can't install without administrator permissions (Windows):**  
Download the **User Installer** variant from the VS Code Downloads page — this installs per-user and does not require administrator privileges.

**`code` command not found in terminal:**
- *Windows:* Open System Properties → Advanced → Environment Variables, and verify the VS Code `bin` directory is in the `PATH`.  
  Run: `rundll32 sysdm.cpl,EditEnvironmentVariables`
- *macOS / Linux:* Open the Command Palette and run **"Shell Command: Install 'code' command in PATH"**.

For manual PATH configuration:
- Mac: `https://code.visualstudio.com/docs/setup/mac`
- Linux: `https://code.visualstudio.com/docs/setup/linux`
- Windows PATH guide: `https://stackoverflow.com/questions/9546324`

**VS Code says installation is corrupt:**  
Usually caused by a theme that modifies product files. Install the **"Fix VS Code Checksums"** extension from the marketplace to apply a fix.

**Need more help:**
- FAQ: `https://code.visualstudio.com/Docs/supporting/faq`
- StackOverflow: `https://stackoverflow.com/questions/tagged/visual-studio-code`

## Module 02: Installing Node.js

Node.js is a JavaScript runtime built on Chrome's V8 engine. It lets you run JavaScript outside the browser — on servers, in build tools, and in the terminal. This module covers installing Node.js and npm, verifying the installation, and troubleshooting common problems.

### What Is Node.js?

Node.js allows you to:

- Run JavaScript programs from the command line
- Build server-side web applications
- Use npm — the world's largest package registry — to manage dependencies
- Power build toolchains (bundlers, transpilers, linters, test runners)

### Installing Node.js

Download from [nodejs.org](https://nodejs.org). Two release lines exist:

| Channel | Description |
|---|---|
| **LTS** (Long-Term Support) | Stable, recommended for most projects |
| **Current** | Latest features, may have breaking changes |

For production or learning projects, prefer the LTS release.

### Verifying the Installation

After installing, open a terminal and run:

```bash
node -v
npm -v
```

Both commands should print version numbers. If either fails, see the troubleshooting section below.

### Running JavaScript with Node.js

**REPL (interactive mode):**

```bash
node
> console.log("Hello World!")
Hello World!
undefined
```

**Running a file:**

```bash
ls
cat index.js
node index.js
```

### Environment Variables in Node.js

Access environment variables in code via `process.env`:

```js
process.env.HOME        // home directory
process.env.GITHUB_TOKEN
process.env.PATH
process.env.NODE_ENV    // 'development' or 'production'
```

Set `NODE_ENV` for the current command:

```bash
# macOS / Linux
NODE_ENV=production npm start

# Windows PowerShell — the above syntax does NOT work
# Use cross-env package or set separately:
$env:NODE_ENV="production"; npm start
```

The error on Windows looks like:
```
The term 'NODE_ENV=production' is not recognized as the name of a cmdlet...
```
Install the [`cross-env`](https://www.npmjs.com/package/cross-env) package to write cross-platform `NODE_ENV` assignments.

### Troubleshooting Node.js

**`node` is not recognized (Windows):**  
```
'node' is not recognized as an internal or external command
```
The Node.js `bin` directory is not on your `PATH`. Re-run the installer and check "Add to PATH", or manually add it via System Properties → Environment Variables.

**Too many open files error:**
```
Error: EMFILE, too many open files
```
Check your current file descriptor limit:
```bash
sysctl -A | grep kern.maxfiles
```
Raise the limit as needed in your shell profile.

**JavaScript heap out of memory:**
```
FATAL ERROR: (...) Allocation failed - JavaScript heap out of memory
```
Increase Node's memory limit before running your script:
```bash
# 2 GB machine
export NODE_OPTIONS="--max-old-space-size=1536"
# 4 GB machine
export NODE_OPTIONS="--max-old-space-size=2048"
# 8 GB machine
export NODE_OPTIONS="--max-old-space-size=4096"
```

### Summary

> npm is the package manager that ships with Node.js. Once Node.js is installed and `node`/`npm` are on your PATH, you're ready to initialise projects and install packages.

## Module 03: Managing Packages with npm

npm (Node Package Manager) is the default package manager for Node.js and the largest software registry in the world. This module covers the full package lifecycle: initialising projects, installing dependencies, updating, removing packages, understanding semantic versioning, and troubleshooting common errors.

### Why npm?

Before package managers, third-party scripts were referenced directly in HTML:

```html
<script type="text/javascript" src="lib/jquery.min.js"></script>
<script type="text/javascript">
(function () {
  const container = $('.container');
})();
</script>
```

With npm and a module system, you declare dependencies in code instead:

```js
const $ = require('jquery');
const container = $('.container');
```

npm downloads the package automatically — no manual file management.

### How npm Works

````python
render_diagram(Path("assets/01-how-npm-works.png"))
````

When you run `npm install`, npm reads `package.json`, contacts the npm Registry to download the listed packages, extracts them into `node_modules/`, and records the exact resolved versions in `package-lock.json`.

### The Heart of a JavaScript Project: `package.json`

`package.json` is the manifest for a JavaScript project. It describes:

```json
{
  "name": "package-name",
  "version": "1.0.0",
  "dependencies": {
    "express": "5.x"
  },
  "devDependencies": {
    "eslint": "^8.0.0"
  }
}
```

- `name` / `version` — package metadata
- `dependencies` — runtime dependencies (installed in production)
- `devDependencies` — development-only dependencies (linters, test runners, bundlers)
- Each key is a **package ID**; each value is a **version constraint**

### The Nervous System: `package-lock.json`

`package-lock.json` records the exact resolved version of every dependency (including transitive dependencies). It ensures every team member and CI environment installs the exact same versions.

**Commit `package-lock.json`** to source control — this guarantees reproducible builds.

### Core npm Commands

```bash
npm install                  # Install all dependencies from package.json
npm install <package>        # Install a specific package (runtime dependency)
npm update <package>         # Update a package to the highest wanted version
npm uninstall <package>      # Remove a package
```

### Initialising a New Project

```bash
npm init
```

npm will prompt for project metadata:

```
package name: (<folder name>)
version: (1.0.0)
description:
entry point: (index.js)
test command:
git repository:
keywords:
author:
```

This creates a `package.json` in the current directory.

### Installing Dependencies

**Install the latest version (runtime dependency):**
```bash
npm install yargs
npm install yargs --save    # older npm: --save was needed to write to package.json
```
Modern npm (v5+) automatically saves to `dependencies` without `--save`.

**Install a specific version:**
```bash
npm install <package>@<version>

# Latest matching a major
npm install express@5

# Exact version
npm install express@5.1.0
```

**Install a development dependency:**
```bash
npm install <package> --save-dev
npm install <package> -D
```

Dev dependencies go into `devDependencies`. They:
- Are **not installed** in a production Node.js environment
- Help teammates identify which packages are not used at runtime

**Install all project dependencies from an existing `package.json`:**
```bash
npm install
```

### Semantic Versioning (SemVer)

````python
render_diagram(Path("assets/02-semantic-versioning.png"), width=700)
````

Version numbers follow the format `Major.Minor.Patch`:

| Part | Meaning |
|---|---|
| **Major** | Breaking change — existing code may need updates |
| **Minor** | New features, backwards-compatible |
| **Patch** | Bug fixes, backwards-compatible |

**Version constraint syntax in `package.json`:**

| Constraint | Meaning |
|---|---|
| `"5.x"` | Any version with major 5 |
| `"^8.0.0"` | Any version with major 8 (minor and patch can increase) |
| `"~8.0.0"` | Any version with major 8 and minor 0 (only patch can increase) |
| `"8.0.0"` | Exactly version 8.0.0 |

**Check available versions:**
```bash
npm view express versions
# [ '0.14.0', '1.0.0', '4.5.1', '5.1.1', '5.2.6', '5.6.0', '5.7.0' ]
```

### Package Version Matching

When `package.json` says `"express": "5.x"`, npm selects the latest published version matching that constraint. If a newer version like `5.7.0` is published after your last install, `npm update` will pick it up — but `npm install` alone will not (it respects `package-lock.json`).

### Listing and Updating Packages

**See which packages have updates available:**
```bash
npm outdated
```
Displays: package name, current version, wanted version (matching constraint), and latest version.

**Update all packages to the highest *wanted* version:**
```bash
npm update
```

> **Important:** `npm update` only updates to the highest **wanted** version (constrained by your version string), not the absolute latest.

**Dry run — preview what would change:**
```bash
npm update --dry-run
```

**Update to the absolute latest version:**
```bash
npm install <package>@latest
```
This overwrites the version constraint in `package.json`.

### Removing a Package

```bash
npm uninstall chalk
```

Completely removes the package from `node_modules` and updates both `package.json` and `package-lock.json`.

### How IDEs Integrate with npm

````python
render_diagram(Path("assets/03-ide-npm-integration.png"))
````

VS Code integrates with npm in six ways:

1. **Package dependency management** — the NPM SCRIPTS panel and package.json editor
2. **Code completion for modules** — `require('...')` completions from installed packages
3. **Language tooling versions and configuration** — reads `package.json` for eslint/prettier versions
4. **Automatic type acquisition** — downloads type definitions from DefinitelyTyped
5. **Detect problems and errors** — highlights errors in the Problems panel
6. **Run npm scripts and tasks** — execute scripts directly from the NPM SCRIPTS panel

#### jsconfig.json — Configuring IDE Tooling

`jsconfig.json` tells VS Code how to interpret your JavaScript project:

```json
{
  "compilerOptions": {
    "module": "CommonJS",
    "moduleResolution": "Node",
    "target": "ES2015"
  },
  "typeAcquisition": {
    "include": ["node"]
  },
  "exclude": [
    "node_modules",
    "**/node_modules/*"
  ]
}
```

- `module` / `moduleResolution` — tells the IDE which module system is in use (CommonJS for Node.js)
- `typeAcquisition.include: ["node"]` — automatically downloads Node.js type definitions, enabling IntelliSense for built-in Node APIs and installed npm packages
- `exclude` — prevents the IDE from indexing `node_modules`

With type acquisition enabled, VS Code shows autocompletion and inline documentation for packages like `yargs` even though they're plain JavaScript.

> **You aren't required to fix IDE errors, but they may anticipate runtime errors.** IDE warnings are advisory — the code still runs — but they often surface real problems before you execute.

### Troubleshooting npm

**Peer dependency error:**
```
ERESOLVE unable to resolve dependency tree
```
Some packages require other packages to be at a certain version range. Workarounds:
```bash
npm install <package> --force
npm install <package> --legacy-peer-deps
```
Prefer `--legacy-peer-deps` as `--force` can silently break things.

**Cannot connect to npm Registry:**
```bash
# Create .npmrc file and set registry
> .npmrc
registry=https://npmjs.com
```
Configure any proxy or firewall settings. See the npm docs for proxy configuration.

**node_modules in an inconsistent state:**
```bash
rm -rf node_modules
npm install
```

**HTTP error when pulling down a package:**
```
code EPROTO
errno EPROTO
request to https://registry.npmjs.org/package failed, reason: write EPROTO...
```
```bash
npm cache clean --force
npm cache verify
```
Reference: `https://stackoverflow.com/a/48826188`

## Module 04: Debugging with Visual Studio Code

VS Code has a built-in debugger that can attach to Node.js processes, Chrome/Edge browser tabs, and arbitrary runtimes via the Debug Adapter Protocol. This module covers the VS Code debugger interface, configuring `launch.json`, debugging both server-side and client-side JavaScript, and using common debugging techniques.

### Why Use a Debugger?

`console.log` debugging is common but limited — it requires modifying source code, cannot pause execution, and doesn't let you inspect the full program state. A proper debugger lets you:

- **Set breakpoints** — pause execution at any line
- **Step through code** — execute one statement at a time
- **Inspect variables** — see the current value of every variable in scope
- **Watch expressions** — track specific values as they change
- **Examine the call stack** — understand how you arrived at the current line

### The VS Code Debugger Interface

Open the debugger with `Ctrl+Shift+D` (Windows/Linux) or `Cmd+Shift+D` (macOS).

Key panels:
- **Variables** — shows local and global variables in the current scope
- **Watch** — expressions you have manually added to monitor
- **Call Stack** — the sequence of function calls leading to the current line
- **Breakpoints** — list of all active breakpoints

Debug toolbar controls:
- **Continue** (F5) — run until the next breakpoint
- **Step Over** (F10) — execute the current line, do not step into function calls
- **Step Into** (F11) — enter the function being called
- **Step Out** (Shift+F11) — finish the current function and return to the caller
- **Restart** — restart the debug session
- **Stop** — end the debug session

### Configuring the Debugger: `launch.json`

Debug configurations live in `.vscode/launch.json`. VS Code will offer to create this file when you first try to start debugging. 

**Debugging a client-side app (Chrome):**

```json
{
  "version": "0.2.0",
  "configurations": [
    {
      "type": "chrome",
      "request": "launch",
      "name": "Launch Chrome against localhost",
      "url": "http://localhost:5173",
      "webRoot": "${workspaceFolder}"
    }
  ]
}
```

This configuration:
1. Launches a new Chrome instance
2. Opens the app at `http://localhost:5173` (a Vite dev server URL)
3. Maps the `webRoot` so source maps resolve correctly

Start your dev server first (`npm run dev`), then press **F5** to attach the debugger.

### Setting Breakpoints

Click in the gutter to the left of a line number, or press `F9` while the cursor is on the line. A red circle appears to confirm the breakpoint is set.

```js
console.log(someVar);  // ← set a breakpoint here to inspect someVar
```

When execution reaches the breakpoint, VS Code pauses, highlights the line, and populates the Variables panel with the current scope.

### Debugging Node.js (Server-Side)

For Node.js scripts, use the **Node.js** debugger type:

```json
{
  "type": "node",
  "request": "launch",
  "name": "Run Node.js script",
  "program": "${workspaceFolder}/index.js"
}
```

Or simply press F5 when a `.js` file is open and no `launch.json` exists — VS Code will auto-create a Node configuration.

### Debugging Client-Side JavaScript

For browser-based apps, VS Code needs either the **JavaScript Debugger** (built-in, works with Edge) or the **Chrome Debugger** extension.

Workflow:
1. Start the development server: `npm run dev` (Vite, webpack-dev-server, etc.)
2. Launch the Chrome debug configuration from VS Code (F5)
3. A controlled Chrome instance opens — breakpoints in VS Code now pause the browser
4. Use the VS Code Variables panel to inspect DOM state, event payloads, etc.
