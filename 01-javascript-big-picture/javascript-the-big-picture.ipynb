### Characteristics of JavaScript

JavaScript is a **dynamic**, **interpreted**, **scripting** language. Understanding each of these terms clarifies how it differs from statically typed, compiled languages like C#.

#### Dynamic vs. Static Typing

In a dynamically typed language, variables do not have fixed types — the type is determined at runtime. Compare:

**JavaScript** (dynamic):
```javascript
// No type declaration needed
let firstName = "David";
let age = 41;
```

**C#** (static):
```csharp
// Type must be declared
string firstName = "David";
int age = 41;
```

#### Compiled vs. Interpreted

A **compiler** translates source code to a lower-level form before execution — like translating a document into a new language all at once. An **interpreter** reads and executes source code line by line at runtime — like translating verbally as you go.

JavaScript is interpreted: the engine reads and runs source code directly. There is no separate compilation step from the developer's perspective.

#### The Complete Definition

Putting it together:

> JavaScript is a **dynamic interpreted scripting language** that can be used to create web applications, backend services, games, mobile applications, and just about anything else you can imagine.

- **Dynamic** — variables and types are resolved at runtime
- **Interpreted** — executed directly by an engine without a separate compilation step
- **Scripting** — designed to automate tasks and run within a host environment (browser, Node.js, etc.)

#### The Three Pillars of Browser Development

Browser-based web applications are built with three complementary technologies:

**HTML — HyperText Markup Language**

HTML defines the structure of a web page. It is the skeleton that organises content into headings, paragraphs, links, images, and other elements.

```html
<!DOCTYPE html>
<html>
  <head>
    <title>HTML Example</title>
  </head>
  <body>
    <p>This is a paragraph. You can define links like
      <a href="https://developer.mozilla.org">MDN</a>.</p>
  </body>
</html>
```

**CSS — Cascading Style Sheets**

CSS controls how HTML looks when rendered. It applies colours, fonts, layout, and spacing.

```css
body {
  background-color: white;
  margin-top: 10px;
}
p {
  color: red;
  font-size: 40px;
}
a {
  text-decoration: none;
}
```

**JavaScript**

JavaScript adds behaviour — responding to user actions, manipulating the DOM, and making the page dynamic.

```html
<head>
  <script>
    function onClick() {
      const m = document.getElementById("menu");
      if (m.style.display === 'none') {
        m.style.display = 'block';
      } else {
        m.style.display = 'none';
      }
    }
  </script>
</head>
<body>
  <button onclick="onClick()">Toggle Menu</button>
  <div id="menu">Site Menu</div>
</body>
```

## Module 03: How Is JavaScript Versioned?

JavaScript has a formal specification and a versioning process managed by an international standards body. Understanding this process explains why new language features appear, how browsers adopt them, and what tools like polyfills and transpilers do.

### Who Governs JavaScript?

**Ecma International** is an international standards organisation. It manages specifications for a wide range of technologies — from CD-ROM file structures to office document formats to programming languages.

The JavaScript specification — called the **ECMAScript specification** — is maintained by Ecma through the **TC-39 committee** (Technical Committee 39).

> "The dot-com boom had begun and Java was the hot new language, so Eich considered the JavaScript name a marketing ploy by Netscape." — Wikipedia

The name "JavaScript" was chosen for marketing reasons. The actual standard is called ECMAScript (often abbreviated ES).

### JavaScript Releases

The ECMAScript specification has gone through several major versions. Each release added new capabilities to the language.

````python
render_diagram(Path("assets/10-ecmascript-timeline.png"))
````

#### Key Releases Summary

| Year | Version | Significance |
|---|---|---|
| 1997 | ES1 | Initial specification |
| 1998 | ES2 | Editorial changes |
| 1999 | ES3 | Try/catch, regular expressions |
| 1999–2008 | ES4 | Proposed but abandoned — dispute over the future of JS |
| 2009 | ES5 | Many new capabilities (strict mode, Array methods, JSON) |
| 2015 | ES6 / ES2015 | Modern JavaScript foundation (classes, modules, arrow functions, Promises) |
| 2017 | ES2017 | async/await |

From ES2015 onward, the committee switched to **annual releases** (ES2016, ES2017, …), each named after the year.

### Extending JavaScript

Not all browsers support the latest ECMAScript features immediately. Two key tools bridge the gap between modern code and older environments.

#### Polyfills

> A polyfill is a piece of code used to provide modern functionality on older browsers that do not natively support it… The reason why polyfills are not used exclusively is for better functionality and better performance. Native implementations of APIs can do more and are faster than polyfills. — Mozilla MDN

A polyfill is a JavaScript implementation of a browser API that is not yet natively available. It adds the missing functionality at runtime.

#### Transpilers

> A transpiler is a translator that takes the source code of a program written in a programming language as its input and produces an equivalent source code in the same or a different programming language. — Wikipedia

A transpiler converts modern JavaScript (e.g. ES2022) to an older dialect (e.g. ES5) so that older browsers can execute it. **Babel** is the most widely used JavaScript transpiler.

**Polyfill vs. Transpiler:**

| Tool | What it does | Example |
|---|---|---|
| Polyfill | Adds missing runtime APIs | `Array.prototype.includes` for IE |
| Transpiler | Converts syntax to an older form | Arrow functions → `function` expressions |

## Module 04: Next Steps with JavaScript

A recap of the key concepts covered and pointers for where to go next in your JavaScript learning path.

### What We Covered

**JavaScript is a dynamic interpreted scripting language** used across a wide range of environments:

- **Browser** — HTML, CSS, and JavaScript together power web UIs; frameworks like React, Angular, Vue, and Svelte build on this foundation
- **Server** — Node.js enables JavaScript on the backend; Express is the dominant web framework
- **Full-stack** — Using one language across the entire stack reduces context-switching and broadens your perspective
- **Beyond the web** — Desktop apps (Electron), mobile (React Native), IoT, and cloud infrastructure tooling

**ECMAScript versioning:**

- The TC-39 committee at Ecma International governs the specification
- ES5 (2009) modernised the language; ES6/ES2015 (2015) introduced the features that define modern JavaScript
- Annual releases since 2015 mean steady, incremental improvements
- Polyfills and transpilers (Babel) let you use modern syntax in environments that have not yet adopted the latest spec

### Suggested Next Steps

1. **JavaScript fundamentals** — variables, functions, control flow, objects, arrays
2. **Modern ES6+ syntax** — arrow functions, destructuring, spread, modules, Promises, async/await
3. **Browser APIs** — DOM manipulation, fetch, events
4. **Node.js and npm** — building backend services and working with packages
5. **A frontend framework** — React is the most in-demand; Vue and Svelte are approachable alternatives
6. **TypeScript** — adds static types to JavaScript; widely used in enterprise and large codebases
