### Deciding on a Runtime for a JavaScript App

#### What Is a Runtime?

A **runtime** is a program or environment that can execute code. For JavaScript, a runtime interprets and executes JS code and provides features and capabilities for it. Think of it as the layer between your code and the hardware or software that runs it — translating your code into instructions the underlying system can understand and execute. The runtime also handles errors, memory management, and security.

**Node.js runtime** — allows JavaScript to run on servers, with its own set of objects and functions for creating web apps, accessing databases, and working with files.

### HTML, DOM, CRUD, and Forms

#### HTML

**HTML** (HyperText Markup Language) is the building block of webpages. It defines the structure and content of a site using special tags (`<tag>`) that browsers interpret and render. Tags define elements such as headings, paragraphs, images, and links.

#### DOM

The **DOM** (Document Object Model) is a programming interface for web documents. It represents the structure of an HTML or XML document as a tree — each element is a "node". Developers use DOM manipulation in JavaScript to interact with and modify the content and structure of a page dynamically, enabling updates and interactivity without full page reloads.

#### CRUD

**CRUD** is the four fundamental data operations:

| Operation | Meaning |
|---|---|
| **Create** | Add new records to the database |
| **Read** | Retrieve existing data |
| **Update** | Modify existing data |
| **Delete** | Remove data |

#### HTML Forms

HTML forms collect user input (text, checkboxes, radio buttons, etc.) via form elements like `<input>`, `<textarea>`, and `<select>`. JavaScript enhances forms by validating input, handling submissions, and providing interactive experiences. On submission, data is sent to the server (back end) for processing.

### Displaying a List with JavaScript

The joke list view uses `fetch()` to call the REST API and dynamically builds `<li>` elements for each joke. Each item includes inline Edit and Delete buttons that call the corresponding CRUD functions:

```javascript
function displayJokes() {
  fetch('/japi/jokes')
    .then(response => response.json())
    .then(data => {
      const jokeList = document.getElementById('joke-list');
      jokeList.innerHTML = '';
      data.forEach(joke => {
        const listItem = document.createElement('li');
        listItem.className = 'list-group-item';
        listItem.innerHTML = `
          <strong>Question:</strong> ${joke.question}<br>
          <strong>Answer:</strong> ${joke.answer}<br>
          <strong>Rating:</strong> ${joke.rating}
          <button onclick="editJoke(${joke.id})">Edit</button>
          <button onclick="deleteJoke(${joke.id})">Delete</button>
        `;
        jokeList.appendChild(listItem);
      });
    });
}
```

### Event Handlers in JavaScript

**Event handlers** are JavaScript code that runs when an action occurs on an HTML or DOM element. They enable the app to respond to user interactions like clicks, page loads, and form submissions.

- A single HTML tag can have multiple event handlers
- Each handler invokes specific code based on the event type
- Handlers can call direct JS expressions or named functions

**Common event handler attributes:**

| Attribute | Triggers when... |
|---|---|
| `onload` | The page or element finishes loading |
| `onclick` | The user clicks the element |
| `onsubmit` | A form is submitted |
| `onchange` | An input value changes |

### JavaScript CRUD Functions and the REST API

#### Entry Point (`index.js`)

The main server file wires together Express, the database connection, static file serving, and the API router:

```javascript
const express = require('express');
const db = require('./config/dbinfo');
const app = express();
const port = 80;
const jokeRouter = require('./apiserver');

app.use(express.json());
app.use(express.static(__dirname));
app.use('/japi', jokeRouter);

app.get('/', (req, res) => {
  res.sendFile(__dirname + '/views/index.html');
});

app.use((req, res) => {
  res.status(404).send('<h1>404 - Not Found</h1>');
});

app.listen(port, () => console.log(`App running on port ${port}`));
```

#### API Routes (`apiserver.js`)

```javascript
const express = require('express');
const db = require('./config/dbinfo');
const router = express.Router();

// Create a joke
router.post('/jokes', (req, res) => {
  let { question, answer, rating } = req.body;
  let sql = 'INSERT INTO jokes (question, answer, rating) VALUES (?, ?, ?)';
  db.query(sql, [question, answer, rating], (err) => {
    if (err) throw err;
    res.send('Joke has been added...');
  });
});

// Update a joke
router.put('/jokes/:id', (req, res) => {
  let { question, answer, rating } = req.body;
  let sql = 'UPDATE jokes SET question=?, answer=?, rating=? WHERE id=?';
  db.query(sql, [question, answer, rating, req.params.id], (err) => {
    if (err) throw err;
    res.send('Joke has been updated...');
  });
});

// Delete a joke
router.delete('/jokes/:id', (req, res) => {
  let sql = `DELETE FROM jokes WHERE id = ${req.params.id}`;
  db.query(sql, (err) => {
    if (err) throw err;
    res.send('Joke has been deleted...');
  });
});

module.exports = router;
```

#### Front-End CRUD Functions

The front-end JavaScript calls the API using `fetch()`. Each operation corresponds to an HTTP method:

**Create** — POST a new joke:

```javascript
async function addJoke(question, answer, rating) {
  try {
    const response = await fetch('/japi/jokes', {
      method: 'POST',
      headers: { 'Content-Type': 'application/json' },
      body: JSON.stringify({ question, answer, rating }),
    });
    if (response.ok) {
      alert('Joke added successfully!');
    } else {
      alert('Failed to add joke.');
    }
  } catch (error) {
    console.log(error);
  }
}
```

**Update** — PUT updated fields:

```javascript
function editJoke(id) {
  const joke = {
    question: prompt('Enter the new question:'),
    answer: prompt('Enter the new answer:'),
    rating: prompt('Enter the new rating:'),
  };
  fetch(`/japi/jokes/${id}`, {
    method: 'PUT',
    headers: { 'Content-Type': 'application/json' },
    body: JSON.stringify(joke)
  })
    .then(response => response.text())
    .then(data => { alert(data); displayJokes(); });
}
```

**Delete** — DELETE by ID:

```javascript
function deleteJoke(id) {
  if (confirm('Are you sure you want to delete this joke?')) {
    fetch(`/japi/jokes/${id}`, { method: 'DELETE' })
      .then(response => response.text())
      .then(data => { alert(data); displayJokes(); })
      .catch(error => console.log(error));
  }
}
```

## Module 04: Publishing Your App

Once built and tested locally, the app needs to be deployed to a hosting environment. This module covers the components required for hosting, the available hosting options across cloud providers, and how CI/CD pipelines automate the release process.

### Components Needed for Hosting

A JavaScript web app has two deployment targets — the front end and the back end — each with its own technology choices:

| Layer | Technologies |
|---|---|
| **Front end** | HTML, CSS, JavaScript; CSS frameworks: Bootstrap, Tailwind |
| **Back end** | Node.js runtime; frameworks: Express, Fastify, Koa |
| **Database** | MySQL or PostgreSQL, hosted on a managed cloud service |

The Dad Jokes app uses Node.js + Express for the back end and Bootstrap for the front-end styling, deployed together as a single full-stack application.

### What Is JavaScript Hosting?

**JavaScript hosting** refers to web hosting that supports JavaScript-based apps running on runtimes like Node.js. Hosting options fall into several categories:

- **Shared servers** — low cost, limited control
- **Virtual private servers (VPS)** — dedicated resources on shared hardware
- **Dedicated servers** — full hardware ownership
- **Containers** — portable, isolated units (Docker + Kubernetes)

For Node.js apps, you need a host that supports the Node.js runtime and can serve it on a public port.

### CI/CD for JavaScript

**CI/CD** (Continuous Integration / Continuous Deployment) is a software engineering practice that automates the path from code commit to production. The pipeline stages are:

1. **Develop** — Code is written and checked into the repository
2. **Build** — A build is triggered and deployed to a test environment
3. **Testing** — Automated tests run to ensure the code has no errors
4. **Package** — The code is packaged for deployment (e.g. as a container image)
5. **Deploy** — The package is deployed to staging and production environments
